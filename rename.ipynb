```python
import os

# Chemin vers le dossier principal du dataset
dataset_path = r"/Users/alaethelegend/Downloads/data2"

# Comptage des images
counts = {}

for split in ["train", "valid", "test"]:
    images_path = os.path.join(dataset_path, split, "images")
    
    if os.path.exists(images_path):
        image_files = [
            f for f in os.listdir(images_path)
            if f.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp'))
        ]
        counts[split] = len(image_files)
    else:
        counts[split] = 0

# Nombre total d'images
total_images = sum(counts.values())

# Affichage des résultats
print(f"Nombre total d'images : {total_images}\n")

for split in ["train", "valid", "test"]:
    percentage = (counts[split] / total_images) * 100 if total_images > 0 else 0
    print(f"{split.capitalize()} :")
    print(f"  Nombre d'images : {counts[split]}")
    print(f"  Pourcentage : {percentage:.2f}%\n")
```
