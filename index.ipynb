```python
from ultralytics import YOLO
import cv2

# Charger le modèle entraîné
model = YOLO("/Users/alaethelegend/Documents/Projet_PFA/runs/detect/Lab/weights/best.pt")

# Coefficient de confiance
CONFIDENCE = 0.5

# Ouvrir la webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Erreur : impossible d'ouvrir la caméra")
    exit()

while True:
    ret, frame = cap.read()

    if not ret:
        print("Erreur : impossible de lire l'image")
        break

    # Prédiction YOLO
    results = model.predict(
        source=frame,
        conf=CONFIDENCE,
        show=False,
        verbose=False
    )

    result = results[0]

    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        label = model.names[class_id]

        print(f"Objet détecté : {label} | Confiance : {confidence:.2f}")

    annotated_frame = result.plot()

    cv2.imshow("Détection en temps réel - YOLO", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
